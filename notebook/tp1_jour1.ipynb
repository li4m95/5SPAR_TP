```python
data = [
    ("John", 23),
    ("Jane", 29),
    ("Dave", 35)
]

df = spark.createDataFrame(
    data,
    ["Name", "Age"]
)

df.show()
```

```python
df.rdd.getNumPartitions()
```

```python
df = df.repartition(2)
df.rdd.getNumPartitions()
```

```python
from pyspark.sql.functions import col

df_filtered = df.filter(
    col("Age") > 25
)
```

```python
df_selected = df_filtered.select(
    "Name",
    "Age"
)
```

```python
df_grouped = df_selected.groupBy("Age").count()
```

```python
df_grouped.show()
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("SparkUIExercise") \
    .getOrCreate()
data = [
    ("Paris", 100),
    ("Paris", 150),
    ("Lyon", 80),
    ("Lyon", 120),
    ("Marseille", 200)
]

df = spark.createDataFrame(
    data,
    ["city", "amount"]
)
df.count()
```

```python
data = [
    ("Paris", "Laptop", 1200),
    ("Paris", "Mouse", 25),
    ("Lyon", "Laptop", 1100),
    ("Lyon", "Keyboard", 80),
    ("Marseille", "Laptop", 1300),
    ("Marseille", "Mouse", 30),
    ("Paris", "Keyboard", 75),
    ("Lyon", "Mouse", 20)
]
df = spark.createDataFrame(
    data,
    ["city", "product", "amount"]
)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
df.filter("amount > 100").show()
```

```python
df.groupBy("city").sum("amount").show()
```

```python
from pyspark.sql import functions as F

df.groupBy("city").agg(F.sum("amount").alias("total")).orderBy(F.col("total").desc()).show()
```

```python
df.rdd.getNumPartitions()
```

```python
df = df.repartition(3)
```

```python
df.count()
```
